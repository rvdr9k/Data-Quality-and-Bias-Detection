Identify Target Column

```python
target_column = 'average_score'
```

Target Balance

```python
df[target_column].value_counts(normalize=True)
```

Separate Numeric & Categorical Columns

```python
numeric_cols = df.select_dtypes(include="number").columns.tolist()
categorical_cols = df.select_dtypes(exclude="number").columns.tolist()

numeric_cols, categorical_cols
```

Feature–Target Relationship Scan

```python
numeric_corr = df[numeric_cols].corr()[target_column].sort_values(ascending=False)
numeric_corr
```

Group Size Check

```python
for col in categorical_cols:
    print(col)
    print(df[col].value_counts(normalize=True).head())
    print("-" * 30)
```

Basic Distribution Sanity

```python
df[numeric_cols].describe().T
```

Identifier & Time Column Detection
