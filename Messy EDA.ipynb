## **Messy Data Analysis**

Import Dataset

```python
import pandas as pd
```

```python
df = pd.read_csv('messy dataset.csv')
```

```python
df.head()
```

```python
df.shape
```

Column name normalization

```python
df.columns = df.columns.str.strip().str.replace("\n", " ", regex=False)
```

Schema & Data types

```python
df.columns.tolist()
```

```python
df.dtypes
```

Target Column 

```python
target_column = 'Average gross'
```

Target Validation

```python
df[target_column].dtype
df[target_column].isna().mean()
df[target_column].value_counts()
```

Missingness Analysis

```python
df.isna().mean()
df.isna().mean(axis=1)
```

Converting Currency columns to Numeric columns 

```python
currency_columns = ['Actual\xa0gross', 'Average gross' , 'Adjusted\xa0gross (in 2022 dollars)']
```

```python
for col in currency_columns:
    df[col] = (
        df[col]
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
    )
    df[col] = pd.to_numeric(df[col], errors="coerce")
```
