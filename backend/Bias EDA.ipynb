Missing Value Percentage

```python
missing_pct = df.isna().mean().sort_values(ascending=False)
missing_pct
```

Rows with Heavy Missingness

```python
row_missing_pct = df.isna().mean(axis=1)
row_missing_pct.describe()
```

Identify Target Column

```python
target_column = 'decision'
```

Target Balance

```python
df[target_column].value_counts(normalize=True)
```

Separate Numeric & Categorical Columns

```python
numeric_cols = df.select_dtypes(include="number").columns.tolist()
categorical_cols = df.select_dtypes(exclude="number").columns.tolist()

numeric_cols, categorical_cols
```

Feature–Target Relationship Scan

```python
group_columns = [ 
  'gender',
  'nationality',
  'sport',
]
```

```python
baseline_rate = df[target_column].mean()
baseline_rate
```

```python
for col in group_columns:
    print(f"\nBias check for: {col}")
    print(df.groupby(col)[target_column].mean())
```

Group Size Check

```python
for col in categorical_cols:
    print(col)
    print(df[col].value_counts(normalize=True).head())
    print("-" * 30)
```

Representation vs Outcome Share

```python
for col in group_columns:
    print(f"\nRepresentation vs outcome for {col}")
    print("Representation:")
    print(df[col].value_counts(normalize=True))
    print("Outcome share:")
    print(df[df[target_column] == True][col].value_counts(normalize=True))

```

Small group risk

```python
df[group_columns].value_counts()
```
